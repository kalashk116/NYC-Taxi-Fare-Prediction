# DROPPING IRRELEVANT COLUMNS

```python
df.drop(columns=['payment_type','passenger_count','trip_type','congestion_surcharge','VendorID','store_and_fwd_flag','RatecodeID','fare_amount','extra','mta_tax','tip_amount','tolls_amount','improvement_surcharge'],inplace=True)
```

```python
df.info()
```

# REPLACING COLUMNS WITH KEY VALUES

```python
df['lpep_pickup_datetime'] = pd.to_datetime(df['lpep_pickup_datetime'])
df['lpep_dropoff_datetime'] = pd.to_datetime(df['lpep_dropoff_datetime'])
```

```python
print(min(df['total_amount']))
print(max(df['total_amount']))

k=[i for i in df['total_amount'] if i < 0]
len(k)

df = df[df['total_amount'] >= 0]
```

```python
# checking neg values in data
k=[i for i in df['total_amount'] if i < 0]
len(k)
```

```python
df["day"] = df["lpep_pickup_datetime"].dt.day
df["month"] = df["lpep_pickup_datetime"].dt.month
df['hour'] = df['lpep_pickup_datetime'].dt.hour
```

```python
df.head(2)
```

```python
df=df[['hour','month','day','PULocationID','DOLocationID','trip_distance','total_amount']]
```

```python
df = df[~df['PULocationID'].isin([264, 265]) & ~df['DOLocationID'].isin([264, 265])]
```

#  ONE-HOT ENCODING 

```python
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder()

df_ohe = pd.get_dummies(df, columns=['PULocationID', 'DOLocationID'], drop_first=False)
df_ohe
```

```python
df_ohe.dropna(inplace=True)
```

```python
df_ohe.to_csv('2020_ohe_predix.csv')
```

# TRAIN-TEST SPLIT

```python
y=df_ohe['total_amount']
x=df_ohe.drop(columns=['total_amount'])
```

```python
from sklearn.model_selection import  train_test_split
x_train, x_test, y_train , y_test = train_test_split(x,y,random_state=7)

print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

# APPLY XGBOOST

```python
import xgboost
xg = xgboost.XGBRegressor()
model = xg.fit(x_train, y_train, verbose=2)

y_pred = model.predict(x_test)
y_pred
```

# MODEL METRICS

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, root_mean_squared_error, mean_absolute_percentage_error

mae = mean_absolute_error(y_test, y_pred)
print("Mean Absolute Error:", mae)

mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)

rmse = root_mean_squared_error(y_test, y_pred)
print("Root Mean Squared Error:", rmse)

r2 = r2_score(y_test, y_pred)
print("R-squared:", r2)
```
